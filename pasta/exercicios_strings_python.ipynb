## 1. Capitalização

Métodos abordados: `upper()`, `lower()`, `capitalize()`, `title()`, `swapcase()`

Esses métodos alteram a forma como letras maiúsculas e minúsculas aparecem em uma string.

### Exercício 1.1

Uma empresa precisa padronizar os nomes dos funcionários em seu sistema. Dado o nome abaixo com capitalização irregular, aplique o método adequado para que apenas a primeira letra de cada palavra fique maiúscula.

```python
nome = "mARIA da silVA sOUZA"
resultado = nome.title()
print(resultado)
```

### Exercício 1.2

Um sistema de autenticação precisa comparar senhas sem diferenciar maiúsculas de minúsculas. Converta as duas strings abaixo para letras minúsculas e verifique se são iguais.

```python
senha_digitada = "Python2024"
senha_armazenada = "python2024"
if senha_digitada.lower() == senha_armazenada.lower():
    print("Senhas conferem!")
else:
    print("Senhas diferentes.")
```

### Exercício 1.3

Um log de sistema registrou uma mensagem toda em maiúsculas. Use o método correto para inverter a capitalização de cada caractere da string.

```python
mensagem = "ERRO CRÍTICO NO SERVIDOR"
resultado = mensagem.swapcase()
print(resultado)
```

## 2. Busca

Métodos abordados: `find()`, `rfind()`, `index()`, `count()`, `startswith()`, `endswith()`

Esses métodos permitem localizar substrings e verificar a presença de padrões dentro de uma string.

### Exercício 2.1

Um sistema de e-mail precisa validar se o endereço digitado pelo usuário contém o símbolo '@' e termina com '.com' ou '.com.br'. Faça essa verificação para o e-mail abaixo.

```python
email = "usuario@empresa.com.br"
tem_arroba = "@" in email
termina_correto = email.endswith(".com") or email.endswith(".com.br")
if tem_arroba and termina_correto:
    print("E-mail válido")
else:
    print("E-mail inválido")
```

### Exercício 2.2

Um analisador de logs precisa contar quantas vezes a palavra 'ERROR' aparece em uma linha de log. Use o método adequado.

```python
log = "ERROR: conexão recusada. Tentativa 1 ERROR. Tentativa 2 ERROR."
quantidade = log.count("ERROR")
print(f"A palavra ERROR aparece {quantidade} vez(es).")  # 3 vezes
```

### Exercício 2.3

Dado um caminho de arquivo, use o método adequado para encontrar a posição da ÚLTIMA ocorrência da barra '/' e extrair apenas o nome do arquivo.

```python
caminho = "/home/usuario/documentos/relatorio_final.pdf"
pos = caminho.rfind("/")
nome_arquivo = caminho[pos + 1:]
print(nome_arquivo)  # relatorio_final.pdf
```

## 3. Modificação

Métodos abordados: `replace()`, `strip()`, `lstrip()`, `rstrip()`, `zfill()`

Esses métodos criam novas strings com alterações no conteúdo ou nos espaços da string original. Lembre-se: strings em Python são imutáveis — os métodos sempre retornam uma nova string.

### Exercício 3.1

Um sistema recebe o CPF do usuário em diferentes formatos. Remova todos os pontos e traços da string abaixo para deixar apenas os números.

```python
cpf = "123.456.789-00"
cpf_limpo = cpf.replace(".", "").replace("-", "")
print(cpf_limpo)  # 12345678900
```

### Exercício 3.2

Um formulário web retornou o nome do usuário com espaços extras no início e no fim. Remova esses espaços antes de salvar no banco de dados.

```python
nome = "   João Silva   "
nome_limpo = nome.strip()
print(f"[{nome_limpo}]")  # [João Silva]
```

### Exercício 3.3

Um sistema de geração de códigos de pedido precisa que todos os números tenham exatamente 8 dígitos, preenchendo com zeros à esquerda quando necessário. Aplique o método adequado.

```python
numero_pedido = "4521"
codigo = numero_pedido.zfill(8)
print(codigo)  # 000004521
```

## 4. Divisão e Junção

Métodos abordados: `split()`, `rsplit()`, `splitlines()`, `join()`

Esses métodos permitem dividir uma string em partes ou unir uma coleção de strings em uma única string.

### Exercício 4.1

Um arquivo CSV contém uma linha com dados de um produto separados por ponto e vírgula. Separe os campos e exiba cada um em uma linha.

```python
linha = "notebook;Dell;i7;16GB;512GB SSD"
campos = linha.split(";")
for campo in campos:
    print(campo)
```

### Exercício 4.2

Uma lista de tags de um sistema de chamados precisa ser exibida como uma string única, com as tags separadas por ' | '. Use o método adequado.

```python
tags = ["urgente", "rede", "servidor", "producao"]
resultado = " | ".join(tags)
print(resultado)  # urgente | rede | servidor | producao
```

### Exercício 4.3

Um script recebe um texto com múltiplas linhas e precisa processá-las individualmente. Divida o texto abaixo em uma lista de linhas e exiba quantas linhas foram encontradas.

```python
texto = """Linha 1: inicio do log
Linha 2: processando
Linha 3: concluido"""
linhas = texto.splitlines()
print(f"Total de linhas: {len(linhas)}")  # 3
for linha in linhas:
    print(linha)
```

## 5. Verificação

Métodos abordados: `isdigit()`, `isalpha()`, `isalnum()`, `isspace()`, `isupper()`, `islower()`

Esses métodos retornam `True` ou `False` e são muito úteis para validar entradas de usuário.

### Exercício 5.1

Um sistema de cadastro precisa validar que o campo 'idade' recebido como string contém apenas dígitos numéricos antes de convertê-lo para inteiro. Faça essa verificação.

```python
idade_input = "25"
if idade_input.isdigit():
    idade = int(idade_input)
    print(f"Idade válida: {idade}")
else:
    print("Entrada inválida: informe apenas números.")
```

### Exercício 5.2

Um validador de nomes de usuário exige que o nome contenha apenas letras e números (sem espaços ou símbolos especiais). Verifique se o nome abaixo é válido.

```python
usuario = "dev_joao_2024"
if usuario.isalnum():
    print("Nome de usuário válido.")
else:
    print("Nome inválido: use apenas letras e números.")  # underscore não é alnum
```

### Exercício 5.3

Um sistema de senha exige que a senha digitada contenha ao menos uma letra maiúscula e ao menos um dígito. Verifique se a senha abaixo atende aos dois critérios usando métodos de verificação de string.

```python
senha = "Segura2024"
tem_maiuscula = any(c.isupper() for c in senha)
tem_digito = any(c.isdigit() for c in senha)
if tem_maiuscula and tem_digito:
    print("Senha válida!")
else:
    print("Senha fraca: inclua maiúsculas e números.")
```

## 6. Formatação

Métodos abordados: `format()`, f-strings, `center()`, `ljust()`, `rjust()`

Esses recursos permitem construir strings dinâmicas e controlar o alinhamento do texto.

### Exercício 6.1

Um sistema de relatório precisa exibir o nome do produto e seu preço em uma linha formatada com f-string, mostrando o preço com duas casas decimais e o símbolo R$.

```python
produto = "SSD 512GB"
preco = 349.9
print(f"Produto: {produto} | Preço: R$ {preco:.2f}")  # R$ 349.90
```

### Exercício 6.2

Um script precisa gerar um cabeçalho centralizado em uma linha de 50 caracteres, preenchendo os lados com o caractere '='. Use o método adequado.

```python
titulo = "RELATÓRIO MENSAL"
cabecalho = titulo.center(50, "=")
print(cabecalho)  # ================RELATÓRIO MENSAL================
```

### Exercício 6.3

Um sistema de inventário precisa exibir uma tabela alinhada no terminal, com o nome do item alinhado à esquerda em 20 caracteres e a quantidade alinhada à direita em 5 caracteres. Formate as duas linhas abaixo.

```python
itens = [("Teclado USB", 15), ("Monitor 24 pol", 4)]
for nome, qtd in itens:
    print(nome.ljust(20) + str(qtd).rjust(5))
```
